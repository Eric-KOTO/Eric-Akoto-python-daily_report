### Exercise 25:
Count the number of words in a text file `"document.txt"`.

```python
# Exercise 25
```

````python
with open("Document.txt", "w") as file:
    file.write("""This is a sample document file.
It contains multiple lines of text.
We will count the number of words in this document.
Each word separated by spaces will be counted.
Let's see how many words we have here!""")

def count_words(filename):
    try:
        with open(filename, "r") as file:
            content = file.read()
            words = content.split()
            return len(words)
    except FileNotFoundError:
        print(f"Error: {filename} not found!")
        return 0
word_count = count_words("Document.txt")
print(f"Number of words in Document.txt: {word_count}")
````

### Exercise 33:
Write a Python function that reads a file and prints the longest line in the file.

```python
# Exercise 33
```

````python
def find_longest_line(filename):
    with open(filename, "r") as file:
        lines = file.readlines()
        return max(lines, key=len)

with open("test_lines.txt", "w") as f:
    f.write("Short\nThis is a much longer line\nMedium")

print("Longest line:", find_longest_line("test_lines.txt").strip())
````

### Exercise 37:
Write a Python function that reads a file and removes all punctuation marks from the text.

```python
# Exercise 37
```

````python
import string

def remove_punctuation(filename):
    with open(filename, "r") as file:
        content = file.read()
        return content.translate(str.maketrans('', '', string.punctuation))

with open("punctuated.txt", "w") as f:
    f.write("Hello, this is text! With punctuation; like this.")

clean_text = remove_punctuation("punctuated.txt")
print("Text without punctuation:", clean_text)
````

### Exercise 39:
Write a Python script to count the number of lines that start with a specific letter (e.g., "A") in a text file.

```python
# Exercise 39
```

````python
def count_lines_starting_with(filename, letter):
    count = 0
    with open(filename, "r") as file:
        for line in file:
            if line.strip().startswith(letter):
                count += 1
    return count

with open("starts_with.txt", "w") as f:
    f.write("Apple is a fruit\nBanana is yellow\nAnt is small\nApple pie is delicious")

letter = "A"
count = count_lines_starting_with("starts_with.txt", letter)
print(f"Lines starting with '{letter}': {count}")
````

### Exercise 43:
Write a Python script to find the most frequent word in a text file `"book.txt"`.

```python
# Exercise 43
```

````python
from collections import Counter

with open("book.txt", "w") as f:
    f.write("the quick brown fox jumps over the lazy dog the fox is quick")

with open("book.txt", "r") as file:
    words = file.read().lower().split()
    most_common = Counter(words).most_common(1)
    print(f"Most frequent word: '{most_common[0][0]}'")
````

### Exercise 48:

```python
# Exercise 48
```

---

````python
from collections import Counter

def most_common_letter(filename):
    with open(filename, "r") as file:
        text = file.read().lower()
        letters = [char for char in text if char.isalpha()]
        return Counter(letters).most_common(1)[0][0]

with open("letter_test.txt", "w") as f:
    f.write("This is a test for counting letters.")

print("Most common letter:", most_common_letter("letter_test.txt"))
````
